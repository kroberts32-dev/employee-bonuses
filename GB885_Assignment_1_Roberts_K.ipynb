## Update Notebook

Create a loop

```python
# Create a loop that calculates the bonuses for all five bank tellers at the South Street Branch.
# Output should include the teller's name and their bonus

for employee in employees:
  employee_name = employee['name']
  employee_rating = employee['rating']
  employee_salary = employee['salary']

  if employee_rating >= 4.5:
    bonus_percentage = 0.17
  elif employee_rating >= 4:
    bonus_percentage = 0.10
  elif employee_rating >= 3:
    bonus_percentage = 0.05
  elif employee_rating >= 1.5:
    bonus_percentage = 0.02
  else:
    bonus_percentage = 0.0

  calculated_bonus = bonus_percentage * employee_salary
  print(f"{employee_name} has earned a bonus of {calculated_bonus}")
```
